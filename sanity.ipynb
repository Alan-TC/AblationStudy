#### Utils

```python
def mean_std_geral(df, columns):
    data = {'mean': [],
            'std': []}
    
    for column in columns:
        data['mean'].append(f'{df[column].mean():.2}')
        data['std'].append(f'{df[column].std():.2}')

    
    return pd.DataFrame(data, index=columns)
```
